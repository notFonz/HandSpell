```python
# Imports
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
import joblib

from tensorflow.keras import Sequential, Input, layers
from tensorflow.keras import optimizers
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.utils import to_categorical
from keras.ops import expand_dims

from sklearn.model_selection import train_test_split
```

```python
# Call dataset_load and dataset_pre
from handspell.model import dataset_load, dataset_preprocessing

train, test = dataset_load('../data/raw_data/A01_kaggle_aslmnist/sign_mnist_train/sign_mnist_train.csv',
                           '../data/raw_data/A01_kaggle_aslmnist/sign_mnist_test/sign_mnist_test.csv')


X_train, y_train, X_test_original, y_test_original = dataset_preprocessing(train, test)
```

```python
# Call validation_split

from handspell.model import validation_split

X_test, X_val, y_test, y_val = validation_split(X_test_original, y_test_original, 0.3, 1)
```

## Model creation, fine-tuning, training

```python
# Call init_model
from handspell.model import init_model

model = init_model(X_train)
```

```python
# Model Summary
model.summary()
```

```python
# Call train_model
from handspell.model import train_model

model, history = train_model(X_train, y_train, X_val, y_val)
```

```python
# Call evaluate_model
from handspell.model import evaluate_model

evaluate_model(X_test, y_test, model)
```

```python
# Call save_model
from handspell.model import save_model

save_model(model, "test.keras")
```
