```python
import requests
from io import BytesIO
from PIL import Image

r = requests.get("http://127.0.0.1:8080/sign/D")
r.raise_for_status()

# Option 1: save to disk
with open("downloaded.png", "wb") as f:
    f.write(r.content)

# Option 2: load in memory with Pillow
img = Image.open(BytesIO(r.content))
img.show()
```

```python
import matplotlib.pyplot as plt
plt.imshow("downloaded.png")
```
